```python
import json
import requests
import pandas as pd
```

```python
# List of character
search_characters = ['R2-D2', 'Darth Vader', 'Godzilla', 'Luke Skywalker', 'Frodo', \
              'Boba Fett', 'Iron Man', 'Jon Snow', 'Han Solo']

# Set url for API
base_url = 'https://swapi.online/api/people/?search='

# Set empty lists to hold character's height and mass
height = []
mass = []


# Loop through each character
for character in search_characters:
    url = base_url + character.replace(" ", "%20")  # Encode spaces

    try:
        response = requests.get(url)
        data = response.json()

        # Check if results are found
        if data.get("result"):
            character_data = data["result"][0]["properties"]
            height.append(character_data.get("height", "unknown"))
            mass.append(character_data.get("mass", "unknown"))
        else:
            print(f"{character} not found in SWAPI.")
            height.append("unknown")
            mass.append("unknown")

    except Exception as e:
        print(f"Error fetching data for {character}: {e}")
        height.append("unknown")
        mass.append("unknown")
```
