```python
# Dependencies
import requests

url = "http://api.worldbank.org/v2/"

add = "lendingtype?format=json"
```

```python
# Get the list of lending types the world bank has
response = requests.get(url + add)

if response.status_code == 200:
    lending_types = response.json()[1]  # The second element contains the data
    for lt in lending_types:
        print(lt["id"])
else:
    print(f"Failed to fetch data: {response.status_code}")
```

```python
lending_counts = {}

for lt in lending_types:
    lt_id = lt['id']
    lt_name = lt['value']
    
    countries_url = f"https://api.worldbank.org/v2/country?lendingType={lt_id}&format=json"
    country_response = requests.get(countries_url)

    if country_response.status_code == 200:
        metadata = country_response.json()[0]
        total_countries = metadata.get('total', 0)
        lending_counts[lt_name] = total_countries
    else:
        lending_counts[lt_name] = "Error fetching data"


```

```python
# Print the number of countries of each lending type
for lt_name, count in lending_counts.items():
    print(f"{lt_name}: {count} countries")
```
