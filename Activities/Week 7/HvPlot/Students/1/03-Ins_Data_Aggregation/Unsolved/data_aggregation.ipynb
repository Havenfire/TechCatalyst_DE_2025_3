# Data Aggregation

```python
import pandas as pd
import numpy as np
import hvplot.pandas
from pathlib import Path
```

## Numerical Aggregation

```python
# Create a DataFrame for prices and zip codes.
houses_df = pd.DataFrame({
    'prices (in thousands))': np.random.randint(100, 1000, 100),
    'zip': np.random.choice([90210, 60606], 100)
})
display(houses_df.head())
display(houses_df.tail())
```

```python
# Filter out only zip codes that equal 90210.
prices_90210 = houses_df.loc[houses_df['zip'] == 90210]
prices_90210.mean()
```

```python
# Run the groupby function on 'zip' without aggregation
houses_df.groupby('zip')
```

```python
#Run the groupby function in 'zip' with aggregation
houses_df.groupby('zip').mean()
```

## Visual Aggregation

```python
# Create the DataFrame to compare AK and CA.
ak_data = np.arange(100)
bias = np.random.uniform(0, 10, 100)
ca_data = 2*ak_data + bias
df = pd.DataFrame({
    "AK": ak_data,
    "CA": ca_data
})
```

```python
# Create a scatter plot to show the relationship between AK and CA.
df.hvplot.scatter(x="AK", y="CA")
```
