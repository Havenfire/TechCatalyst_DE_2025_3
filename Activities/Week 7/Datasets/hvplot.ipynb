```python
sp500 = pd.read_csv('sp500_companies.csv')
sp500
```

```python
sp500_sectors = sp500['Sector'].value_counts()
```

```python
sp500_sectors.plot(kind='bar', rot=45);
plt.show()
```

```python
sp500_sectors.hvplot(kind='bar', rot=45)
```

```python
sp500_sectors.hvplot.bar(rot=45)
```

```python
sp500['Sector'].value_counts().hvplot(kind='bar').opts(
    title="Sector breakdown",
    hover_color='orange',
    color='purple'
)
```

```python
hospital = pd.read_csv('hospital_claims.csv')
```

```python
hospital.head()
```

```python
procedures_552_charge = hospital[hospital['DRG Definition'] == '552 - MEDICAL BACK PROBLEMS W/O MCC']
procedures_552_charge
```

```python
total_payments_by_state = procedures_552_charge.groupby('Provider State')[['Average Total Payments']].sum()
total_payments_by_state
```

```python
total_payments_by_state.hvplot.bar(rot=90)
```

```python
total_payments_by_state.hvplot.bar(rot=90) + total_payments_by_state.sort_values("Average Total Payments").hvplot()
```

```python
sorted_total_payments_by_state = total_payments_by_state.copy()
sorted_total_payments_by_state.index.names = ["Provider State Sorted"]

sorted_total_payments_by_state.sort_values("Average Total Payments").hvplot(rot=45, label="Trend over time") * total_payments_by_state.hvplot.bar(rot=90, label='Avg Medicare Cost')
```

```python
total_payments_by_state.hvplot.bar(rot=90) + total_payments_by_state.sort_values("Average Total Payments").hvplot()
```
